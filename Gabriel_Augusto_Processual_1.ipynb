Fazendo importação do pandas e do seaborn

```python
import pandas as pd
import seaborn as sns
```

Trazendo a base de dados e verificando

```python
base_steam = pd.read_csv('steam_games.csv')
base_steam.head()
```

```python
base_steam.shape
```

Verificando tipos de dados

```python
base_steam.info()
```

Verificando as medidas de tendência central

```python
numericas = base_steam.select_dtypes(include='number')

print("Media:")
display(numericas.mean())

print("\nDesvio Padrão:")
display(numericas.std())

print("\nMediana:")
display(numericas.median())
```

Verificando a correlação entre as variaveis

```python
numericas.corr()
```

```python
sns.heatmap(numericas.corr())
```
